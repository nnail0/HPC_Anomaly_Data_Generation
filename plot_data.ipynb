```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import re

# Dictionary of anomaly abbreviations
anom_dict = {"ML":"memleak",
             "CO":"cpuoccupy",
             "CC":"cachecopy"}


# Function for reading the data into a pandas DataFrame
def read_data(dataset="./dataset",
              app="ExaMiniMD",
              anom1="",
              anom1_params=[],
              anom2="",
              anom2_params=[],
              overlap=False,
              plugin="meminfo",
              print_info=True):
    """
    dataset: The path pointing to the dataset directory
    app: The name of the application for which the data will be read
    anom1: the first anomaly ran, if any, e.g., "cachecopy"
    anom1_params: a list of specific desired parameters for the first anomaly, e.g.,
                  ["cL1", "m0.8"]
    anom2: the second anomaly ran, if any
    anom2_params: a list of specific desired parameters for the second anomaly
    overlap: if two anomalies, then they overlap if True, and not if False
    plugin: the metric plugin to read, e.g., "meminfo"
    print_info: print information if True, and not if false

    """
    
    # Create the data_dir variable to be the path to the data
    data_dir = dataset + "/" + app

    if overlap:
        relationship="overlaps"
    else:
        relationship="separate"
    
    if anom1 != "":
        data_dir += "_" + anom1
    if anom2 != "":
        data_dir += "_" + anom2 + "/" + relationship

    if os.path.isdir(data_dir + "/data"):
        if print_info: print("   - no anomalies\n")
        data_dir += "/data"
    else:
        # Find all sets of data with the given combo of anomalies
        dirs = [i for i in os.listdir(data_dir)]

        # Check for specified anomaly parameters
        for i in range(len(dirs)):
            directory = dirs[i]
            metadata = re.findall(r'ST(\d+)_ET(\d+)((?:_[a-z][^_ ]+)+)', directory)
            param_match = True
            for param in anom1_params:
                if param not in metadata[0][2].split("_"):
                    param_match = False
            for param in anom2_params:
                if param not in metadata[1][2].split("_"):
                    param_match = False
            if param_match:
                break

        if not param_match:
            raise Exception("no data matches the parameters provided")

        # Print information
        if print_info:
            start1 = metadata[0][0]
            end1 = metadata[0][1]
            param1 = metadata[0][2].replace("_","-")
            print("   - " + anom_dict[anom1] + " (" + param1 + ") from t=" + start1 + " to t=" + end1)

        if len(metadata) > 1:
            if print_info:
                start2 = metadata[1][0]
                end2 = metadata[1][1]
                param2 = metadata[1][2].replace("_","-")
                print("   - " + anom_dict[anom2] + " (" + param2 + ") from t=" + start2 + " to t=" + end2 + "\n")
        else:
            if print_info: print()
        
        data_dir += "/" + directory + "/data"

    if plugin == "meminfo":
        plugin = "memory_metrics/meminfo"
    elif plugin == "vmstat":
        plugin = "memory_metrics/vmstat"
    elif plugin == "procstat":
        plugin = "process_metrics/procstat"

    data_dir += "/" + plugin

    data = pd.read_csv(data_dir)

    return data


# Function for plotting the DataFrame to a graph
def plot_data(data, col_name, norm=False):
    """
    data: the pandas DataFrame with the data to plot
    col_name: the name of the column to plot, e.g., "MemFree"
    norm: normalizes the data if True, and not if False
    """
    column = data[col_name].iloc[5:-5] # Remove first and last rows to avoid extreme values
    if norm:
        # Normalize the data
        norm_column = column.copy()
        norm_column = (norm_column - norm_column.min()) / (norm_column.max() - norm_column.min())
        norm_column.plot(legend=True)
    else:
        column.plot(legend=True)


# Change these variables to visualize different data
application = "ExaMiniMD"
anom1 = "CO"
anom1_params = ["u95"]
anom2 = "ML"
anom2_params = ["s10M"]
plugin = "procstat"
norm = True

data_template = read_data(app=application, plugin=plugin, print_info=False)

for i in range(len(data_template.columns)-6):
    print("Application: " + application)
    print("Metric: " + plugin + ", " + data_template.columns[i+6] + "\n")

    print("  Blue Line Anomalies:")
    data = read_data(app=application, plugin=plugin)
    plot_data(data, data.columns[i+6], norm=norm)

    print("  Orange Line Anomalies:")
    data = read_data(app=application, anom1=anom1, anom1_params=anom1_params, plugin=plugin)
    plot_data(data, data.columns[i+6], norm=norm)

    print("  Green Line Anomalies:")
    data = read_data(app=application, anom1=anom1, anom1_params=anom1_params, anom2=anom2, anom2_params=anom2_params, plugin=plugin)
    plot_data(data, data.columns[i+6], norm=norm)

    print("  Red Line Anomalies:")
    data = read_data(app=application, anom1=anom1, anom1_params=anom1_params, anom2=anom2, anom2_params=anom2_params, overlap=True, plugin=plugin)
    plot_data(data, data.columns[i+6], norm=norm)

    plt.show()

    plt.clf()
```
